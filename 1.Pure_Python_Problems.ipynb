## 1. Задача: Проверить, является ли строка правильной скобочной последовательностью.

### Если в строке могут быть только скобки

```python
def is_valid_brackets(s):
    
    # Создаем пустой стек: здесь будем хранить открывающие скобки,
    # которые еще не были закрыты
    stack = []

    # Словарь соответствий:
    # ключ — закрывающая скобка,
    # значение — открывающая скобка, которая ей соответствует
    pairs = {
        ")" : "(",
        "]" : "[",
        "}" : "{"
    }

    # Набор открывающих скобок
    opening_brackets = ("(", "[", "{")

    # Проходим по строке слева направо, символ за символом
    for char in s:

        # Если символ — открывающая скобка,
        # кладем ее в стек
        if char in opening_brackets:
            stack.append(char)

        # Иначе проверяем, если символ — закрывающая скобка
        elif char in pairs:
            
            # Если стек пустой, значит закрывающая скобка пришла раньше,
            # чем появилась соответствующая открывающая
            if not stack:
                return False
                
            # Проверяем, соответствует ли последняя открытая скобка
            # текущей закрывающей скобке
            if stack[-1] != pairs[char]:
                return False

            # Если скобки соответствуют друг другу,
            # удаляем последнюю открытую скобку из стека
            stack.pop()

        # Иначе считаем, что символ — ни открывающая, ни закрывающая скобка, выводим предупреждение и False результат
        else:
            print(f"Ошибка: в строке есть не скобочный символ: {char}")
            return False

    # После прохода по всей строке стек должен быть пустым:
    # если в нем что-то осталось, значит какие-то открывающие скобки не закрылись
    return len(stack) == 0
```

```python
print(is_valid_brackets('()'))       ##True
print(is_valid_brackets('([)]'))     ##False
print(is_valid_brackets('()[]'))     ##True
print(is_valid_brackets('([{}]())')) ##True
print(is_valid_brackets('(a)'))      ##Предупреждение + False
```

### Если в строке могут быть НЕ только скобки

```python
def is_valid_brackets(s):
    
    # Создаем пустой стек: здесь будем хранить открывающие скобки,
    # которые еще не были закрыты
    stack = []

    # Словарь соответствий:
    # ключ — закрывающая скобка,
    # значение — открывающая скобка, которая ей соответствует
    pairs = {
        ")" : "(",
        "]" : "[",
        "}" : "{"
    }

    # Набор открывающих скобок
    opening_brackets = ("(", "[", "{")

    # Проходим по строке слева направо, символ за символом
    for char in s:

        # Если символ — открывающая скобка,
        # кладем ее в стек
        if char in opening_brackets:
            stack.append(char)

        # Иначе проверяем, если символ — закрывающая скобка
        elif char in pairs:
            
            # Если стек пустой, значит закрывающая скобка пришла раньше,
            # чем появилась соответствующая открывающая
            if not stack:
                return False
                
            # Проверяем, соответствует ли последняя открытая скобка
            # текущей закрывающей скобке
            if stack[-1] != pairs[char]:
                return False

            # Если скобки соответствуют друг другу,
            # удаляем последнюю открытую скобку из стека
            stack.pop()

        # Иначе считаем, что символ — ни открывающая, ни закрывающая скобка и пропускаем этот символ
        else:
            continue

    # После прохода по всей строке стек должен быть пустым:
    # если в нем что-то осталось, значит какие-то открывающие скобки не закрылись
    return len(stack) == 0
```

```python
print(is_valid_brackets('()'))       ##True
print(is_valid_brackets('([)]'))     ##False
print(is_valid_brackets('()[]'))     ##True
print(is_valid_brackets('([{}]())')) ##True
print(is_valid_brackets('(a)'))      ##True
```
