```python
import numpy as np

rng = np.random.default_rng(42)
```

**NumPy** - библиотека для работы с массивами чисел: векторами, матрицами, таблицами числовых признаков.

**NumPy** позволяет делать операции сразу над всем массивом, без ручных циклов Python.

**NumPy** нужен для эффективной работы с числовыми массивами и матрицами. Он позволяет выполнять операции векторно, без явных Python-циклов, и является базой для многих библиотек аналитики и машинного обучения.

Зачем это аналитику

1. Быстрее считать на больших данных
NumPy написан на оптимизированном C под капотом, поэтому операции над массивами работают быстрее, чем обычные Python-циклы.
2. Удобно работать с матрицами
Например:
arr.mean(axis=0)  # среднее по столбцам
arr.mean(axis=1)  # среднее по строкам
3. Маски и фильтрация
arr[arr > 0]
выберет только положительные элементы.
4. Это основа pandas, scipy, sklearn
Много аналитических библиотек внутри опираются на NumPy. Даже если ты чаще работаешь в pandas, под ним часто лежат NumPy-массивы.
5. На собеседовании могут просить “без циклов”
Например:
Найти седловую точку в матрице через NumPy без циклов.

## Задача 3: np.where и категории

Дан массив оценок пользователей от 0 до 100:

```python
scores = rng.integers(0, 101, 8)
```

```python
scores = np.array([95, 67, 82, 40, 76, 100, 58, 73])
```

```python
scores
```

Нужно:
1. создать массив passed, где будет 1, если оценка >= 70, иначе 0;
2. создать массив labels, где:
   - "high", если оценка >= 85;
   - "medium", если оценка от 70 до 84;
   - "low", если оценка ниже 70;
3. посчитать, сколько пользователей в каждой категории.

### 1. Создать массив passed

```python
passed = np.where(scores >= 70, 1, 0)
```

```python
passed
```

### 2. Создать массив labels

```python
labels = np.where(scores < 70, 'low', np.where(scores >= 85, 'high', 'medium'))
```

```python
labels
```

### 3. Посчитать, сколько пользователей в каждой категории (в labels)

```python
categories, counts = np.unique(labels, return_counts = True)
```

```python
np.unique(labels)
```

```python
result = dict(zip(categories, counts))
```

```python
result
```

```python
result = {str(category): int(count) for category, count in zip(categories, counts)}
result
```

```python
result = {
    "high": int((labels == "high").sum()),
    "medium": int((labels == "medium").sum()),
    "low": int((labels == "low").sum())
}

result
```

## Задача 5: Седловая точка в матрице

Дана матрица:

```python
A = np.array([
    [7, 8, 9],
    [5, 6, 7],
    [3, 4, 5]
])
```

Нужно найти элемент, который одновременно является:
- минимальным в своей строке;
- максимальным в своем столбце.

Вернуть индексы такого элемента.

```python
row_min_mask = A == A.min(axis = 1, keepdims=True)
```

```python
row_min_mask
```

```python
col_max_mask = A == A.max(axis = 0, keepdims=True)
```

```python
col_max_mask
```

```python
saddle_mask = row_min_mask & col_max_mask
```

```python
indices = np.argwhere(saddle_mask)
```

```python
indices
```

## Задача 8: Broadcasting

**Broadcasting** — это когда NumPy автоматически “растягивает” массивы совместимых размеров.

Broadcasting буквально можно понимать как “трансляция” или “распространение”.

Смысл названия: NumPy как будто распространяет меньший массив на больший, чтобы выполнить операцию поэлементно.

Например:

```python
X = np.array([
    [10, 100],
    [20, 200],
    [30, 300]
])
```

```python
means = X.mean(axis=0)
```

```python
means
```

Когда пишем:

```python
X - means
```

NumPy вычитает: [20, 200] из каждой строки матрицы.

Дан массив покупок клиентов:

```python
purchases = np.array([
    [100, 200, 300],
    [50,  150, 250],
    [400, 100, 500]
])
```

Строки — клиенты, столбцы — категории товаров.

Дан массив скидок по категориям:

```python
discounts = np.array([0.10, 0.20, 0.15])
```

Нужно:
1. применить скидки к каждой покупке;
2. получить новый массив final_prices;
3. посчитать итоговую сумму покупок по каждому клиенту после скидки.

```python
final_prices = purchases * (1 - discounts)
```

```python
final_prices
```

```python
final_prices.sum(axis = 1)
```
