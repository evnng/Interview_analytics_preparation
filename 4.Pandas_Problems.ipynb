## Задача 3: Создание новых колонок

На этом же df создай новые колонки:
    
1. revenue_per_click = revenue / clicked, но если clicked = 0, должно быть 0, а не бесконечность/ошибка.
2. is_buyer = 1, если purchased = 1, иначе 0.
3. age_group:
   - если age < 30 → "young"
   - если age < 50 → "middle"
   - иначе → "senior"
4. client_value:
   - если revenue >= 2000 → "high"
   - если revenue > 0 → "medium"
   - если revenue = 0 → "zero"

```python
df['revenue_per_click'] = np.where(
    df['clicked'] == 0,
    0,
    df['revenue'] / df['clicked']
)
```

```python
df
```

```python
df['is_buyer'] = df['purchased']
```

```python
df
```

```python
df['age_group'] = np.where(
    df['age'] < 30,
    'young',
    np.where(
        df['age'] >= 50,
        'senior',
        'middle'
    )
)
```

```python
df
```

```python
conditions = [
    df['revenue'] >= 2000,
    df['revenue'] == 0
]

choices = [
    "high",
    "zero"
]

df["client_value"] = np.select(
    conditions,
    choices, 
    default = 'medium'
)
```

```python
df
```
